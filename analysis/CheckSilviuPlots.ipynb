```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

%matplotlib widget
plt.rc('text', usetex = True)
plt.rcParams['font.size'] = 16
plt.rcParams['figure.facecolor'] = 'w'
```

```python
material = 'Be'
thickness = '0.5'
path = f'/Users/max/SLAC Fall 2024/LXeSims/{material}_sigmaR_3mm_thickness_{thickness}mm'

cols = 'x y z Px Py Pz t PDGid EventID TrackID ParentID Weight'.split(' ')

df_LXe = pd.read_csv(path + '/Silviu_LXe.csv')
df_Entrance = pd.read_csv(path + '/Silviu_entrance_window.csv')
df_Exit = pd.read_csv(path + '/Silviu_exit_window.csv')
```

```python
pulse_time_sep = 3.3e-3

LXe_subvolume = 2e-9 # m^3
Window_subvolume = 5e-10 # m^3 (0.5 mm window thickness)

print(f"Entrance power loss: {Window_subvolume * df_Entrance['PowerPerVolume'].sum():.2f} W")
print(f"Exit power loss: {Window_subvolume * df_Exit['PowerPerVolume'].sum():.2f} W")
print(f"LXe power loss: {LXe_subvolume * df_LXe['PowerPerVolume'].sum():.2f} W")

print(f"Entrance energy loss: {pulse_time_sep * Window_subvolume * df_Entrance['PowerPerVolume'].sum():.2f} J")
print(f"Exit energy loss: {pulse_time_sep * Window_subvolume * df_Exit['PowerPerVolume'].sum():.2f} J")
print(f"LXe energy loss: {pulse_time_sep * LXe_subvolume * df_LXe['PowerPerVolume'].sum():.2f} J")
```

```python
plt.close()
fig, ax = plt.subplots(3, 1, figsize=(12,12), gridspec_kw={'hspace': 0.5})


ax[0].scatter(1e3 * df_LXe['x'], df_LXe['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[0].set_xlabel(r'$x\ [\mathrm{mm}]$', fontsize=18)

ax[1].scatter(1e3 * df_LXe['y'], df_LXe['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[1].set_xlabel(r'$y\ [\mathrm{mm}]$', fontsize=18)

ax[2].scatter(1e3 * df_LXe['z'], df_LXe['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[2].set_xlabel(r'$z\ [\mathrm{mm}]$', fontsize=18)

for i in range(3):
    ax[i].set_ylabel(r'$\mathrm{Power\ loss/vol.}\ \left[\ \mathrm{W}\cdot \mathrm{m}^{-3}\right]$', fontsize=18)


plt.suptitle('LXe')
plt.show()
```

```python
plt.close()
fig, ax = plt.subplots(3, 1, figsize=(12,12), gridspec_kw={'hspace': 0.5})


ax[0].scatter(1e3 * df_Entrance['x'], df_Entrance['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[0].set_xlabel(r'$x\ [\mathrm{mm}]$', fontsize=18)

ax[1].scatter(1e3 * df_Entrance['y'], df_Entrance['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[1].set_xlabel(r'$y\ [\mathrm{mm}]$', fontsize=18)

ax[2].scatter(1e3 * df_Entrance['z'], df_Entrance['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[2].set_xlabel(r'$z\ [\mathrm{mm}]$', fontsize=18)

for i in range(3):
    ax[i].set_ylabel(r'$\mathrm{Power\ loss/vol.}\ \left[\ \mathrm{W}\cdot \mathrm{m}^{-3}\right]$', fontsize=18)


plt.suptitle(f'{thickness} mm {material} entrance window')
plt.show()
```

```python
plt.close()
fig, ax = plt.subplots(3, 1, figsize=(12,12), gridspec_kw={'hspace': 0.5})


ax[0].scatter(1e3 * df_Exit['x'], df_Exit['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[0].set_xlabel(r'$x\ [\mathrm{mm}]$', fontsize=18)

ax[1].scatter(1e3 * df_Exit['y'], df_Exit['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[1].set_xlabel(r'$y\ [\mathrm{mm}]$', fontsize=18)

ax[2].scatter(1e3 * df_Exit['z'], df_Exit['PowerPerVolume'], marker='o', facecolors='b', s=10, edgecolors='none')
ax[2].set_xlabel(r'$z\ [\mathrm{mm}]$', fontsize=18)

for i in range(3):
    ax[i].set_ylabel(r'$\mathrm{Power\ loss/vol.}\ \left[\ \mathrm{W}\cdot \mathrm{m}^{-3}\right]$', fontsize=18)


plt.suptitle(f'{thickness} mm {material} exit window')
plt.show()
```
